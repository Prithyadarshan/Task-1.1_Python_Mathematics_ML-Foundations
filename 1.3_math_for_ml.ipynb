## 1. Statistics

**Problem:** For the dataset $X = [12, 15, 12, 18, 20, 15, 22, 15]$, find the mean, median, mode,
variance, and standard deviation.

**Worked solution**

Mean: $\bar{x} = \dfrac{12+15+12+18+20+15+22+15}{8} = \dfrac{129}{8} = 16.125$

Sorted data: $12,12,15,15,15,18,20,22$ → Median $= \dfrac{15+15}{2} = 15$

Mode: $15$ appears 3 times → most frequent value.

Variance (population): $\sigma^2 = \dfrac{1}{n}\sum (x_i - \bar{x})^2$

Standard deviation: $\sigma = \sqrt{\sigma^2}$


```python
import numpy as np

X = np.array([12, 15, 12, 18, 20, 15, 22, 15])

mean = X.mean()
median = np.median(X)
values, counts = np.unique(X, return_counts=True)
mode = values[np.argmax(counts)]
variance = X.var()          # population variance (ddof=0)
std_dev = X.std()

print(f"Mean     = {mean}")
print(f"Median   = {median}")
print(f"Mode     = {mode}")
print(f"Variance = {variance:.4f}")
print(f"Std Dev  = {std_dev:.4f}")
```

## 2. Probability — Bayes' Theorem

**Problem:** A diagnostic test for a disease is 95% sensitive (true positive rate) and 90% specific
(true negative rate). The disease prevalence in the population is 2%. If a patient tests positive,
what is the probability they actually have the disease? (This is exactly the kind of reasoning behind
precision/recall trade-offs in medical image classification.)

**Worked solution**

Let $D$ = has disease, $T^+$ = tests positive.

$P(D) = 0.02$, $P(T^+|D) = 0.95$, $P(T^+|\lnot D) = 1 - 0.90 = 0.10$

Bayes' theorem:
$$P(D|T^+) = \frac{P(T^+|D)\,P(D)}{P(T^+|D)\,P(D) + P(T^+|\lnot D)\,P(\lnot D)}$$


```python
P_D = 0.02
P_pos_given_D = 0.95
P_pos_given_notD = 0.10
P_notD = 1 - P_D

P_pos = P_pos_given_D * P_D + P_pos_given_notD * P_notD
P_D_given_pos = (P_pos_given_D * P_D) / P_pos

print(f"P(positive test) = {P_pos:.4f}")
print(f"P(disease | positive test) = {P_D_given_pos:.4f}  ({P_D_given_pos*100:.1f}%)")
print("\nInterpretation: even with a fairly accurate test, low prevalence means most")
print("positive results are still false positives — a key reason imbalanced medical")
print("datasets need more than accuracy as a metric.")
```

## 3. Linear Algebra

**Problem A:** Multiply matrices $A$ (2x3) and $B$ (3x2).
**Problem B:** Find the eigenvalues and eigenvectors of $C = \begin{bmatrix}4 & 2\\1 & 3\end{bmatrix}$.
**Problem C:** Compute the L2 norm of vector $v = [3, 4]$ and the cosine similarity between
$v_1=[1,2,3]$ and $v_2=[2,4,6]$.


```python
# A. Matrix multiplication
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8],
              [9, 10],
              [11, 12]])
product = A @ B
print("A @ B =\n", product)

# B. Eigenvalues / eigenvectors
C = np.array([[4, 2],
              [1, 3]])
eigvals, eigvecs = np.linalg.eig(C)
print("\nEigenvalues of C:", eigvals)
print("Eigenvectors of C:\n", eigvecs)

# C. Norm and cosine similarity
v = np.array([3, 4])
l2_norm = np.linalg.norm(v)
print(f"\nL2 norm of {v} = {l2_norm}")

v1 = np.array([1, 2, 3])
v2 = np.array([2, 4, 6])
cos_sim = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
print(f"Cosine similarity between {v1} and {v2} = {cos_sim:.4f}  (parallel vectors -> 1.0)")
```

## 4. Calculus — Gradients used in ML

**Problem A:** Derive the derivative of the sigmoid function
$\sigma(x) = \dfrac{1}{1+e^{-x}}$, used as an activation function.

**Worked solution**

$$\sigma'(x) = \sigma(x)\,(1-\sigma(x))$$

**Problem B:** Perform one manual step of gradient descent for the simple loss
$L(w) = (w-3)^2$ starting at $w_0 = 0$ with learning rate $\eta = 0.1$.

**Worked solution**

$\dfrac{dL}{dw} = 2(w-3)$. At $w_0=0$: gradient $= 2(0-3) = -6$.

Update: $w_1 = w_0 - \eta \cdot \text{gradient} = 0 - 0.1 \times (-6) = 0.6$


```python
# A. Verify sigmoid derivative numerically
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative_analytic(x):
    s = sigmoid(x)
    return s * (1 - s)

def numerical_derivative(f, x, h=1e-6):
    return (f(x + h) - f(x - h)) / (2 * h)

x_test = 1.5
analytic = sigmoid_derivative_analytic(x_test)
numeric = numerical_derivative(sigmoid, x_test)
print(f"Analytic sigmoid'({x_test}) = {analytic:.6f}")
print(f"Numeric  sigmoid'({x_test}) = {numeric:.6f}  (should match closely)")

# B. One step of manual gradient descent, then run several steps to see convergence
def loss(w):
    return (w - 3) ** 2

def grad(w):
    return 2 * (w - 3)

w, lr = 0.0, 0.1
print(f"\nGradient descent on L(w) = (w-3)^2, starting w={w}, lr={lr}")
for step in range(6):
    g = grad(w)
    w = w - lr * g
    print(f"  step {step+1}: gradient={g:.4f}, w -> {w:.4f}, loss={loss(w):.4f}")
print("\nw is converging toward the true minimum at w=3.")
```
